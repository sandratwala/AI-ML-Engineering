### 8. Missing Value Imputation
We handle empty data entries to prevent mathematical errors in our AI model:
*   **`.fillna()`**: Replaces missing (`NaN`) fields with a value we specify.
*   **`.median()`**: Finds the middle value of numeric columns (`Age`, `Fare`) to avoid outlier distortion.
*   **`.mode()[0]`**: Finds the most common text value for the categorical column (`Embarked`).

**Why:** Machine learning algorithms will crash or fail if they encounter empty values. Filling them with medians and modes maintains data integrity.


```python
x['Age']=x['Age'].fillna(x['Age'].median())
x['Fare']=x['Fare'].fillna(x['Fare'].median())
x['Embarked']=x['Embarked'].fillna(x['Embarked'].mode()[0])
```

### 9. Encoding Categorical Text Variables
We convert descriptive text columns into a numerical format:
*   **`pd.DataFrame()`**: Initializes an empty data canvas named `train`.
*   **`LabelEncoder()`**: A preprocessing tool that translates text strings into unique integers (e.g., "male" to 1, "female" to 0).

**Why:** AI algorithms only understand mathematical equations and numbers. They cannot directly process raw text fields like names, genders, or ports.


```python
train=pd.DataFrame()
label=LabelEncoder()
for c in x.columns:
    if(x[c].dtype=='object'):
        train[c]=label.fit_transform(x[c])
    else:
        train[c]=x[c]
```

### 18. Feature Scaling and Updated Logistic Regression Pipeline
We implement a data normalization process to improve algorithm stability and performance:
*   **`StandardScaler()`**: A scaling tool that shifts and scales numerical columns to have a mean of 0 and variance of 1.
*   **`.fit_transform()`**: Programmatically calculates column means and standard deviations, then transforms features into scaled arrays.

**Why:** Numerical values like `Ticket` and `PassengerId` contain vastly different magnitudes compared to `Sex` or `Pclass`. Normalization scales all numbers to an identical range, helping gradient descent models converge efficiently and prevent large numbers from dominating the mathematical weight calculations.


```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler

def logistic(x, y):
    scaler = StandardScaler()
    x_scaled = scaler.fit_transform(x)

    x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, random_state=42, test_size=0.2)

    lr = LogisticRegression(max_iter=1000)
    lr.fit(x_train, y_train)
    y_pre = lr.predict(x_test)
    print('Accuracy: ', accuracy_score(y_test, y_pre))


```
